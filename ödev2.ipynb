```python
def max_in_list(nums):
    # varsayım: liste boş değil
    maksimum = nums[0]
    for n in nums[1:]:
        if n > maksimum:
            maksimum = n
    return maksimum

# Örnekler
print(max_in_list([1,2,3,4,5]))  
print(max_in_list([-1,-5,-3]))   
```

```python
def reverse_string(s):
    return s[::-1]

# Örnekler
print(reverse_string("Aslınur"))   
 
```

```python
def factorial(n):
    # n >= 0 varsayımı
    result = 1
    for i in range(2, n+1):
        result *= i
    return result

# Örnekler
print(factorial(5))  # 120
print(factorial(0))  # 1

```

```python
def filter_even(nums):
    return [x for x in nums if x % 2 == 0]

# Örnekler
print(filter_even([1,2,3,4,5,6]))  # [2,4,6]
print(filter_even([11,23,44]))     # [44]

```

```python
def union_unique(a, b):
    return list(set(a) | set(b))

# Örnekler (sıra önemli değil)
print(sorted(union_unique([1,2,3], [3,4,5])))  # [1,2,3,4,5]
print(sorted(union_unique([], [1,1,2])))       # [1,2]
```

```python
import re

def is_palindrome(s):
    cleaned = re.sub(r'[^A-Za-z0-9]', '', s).lower()
    return cleaned == cleaned[::-1]

# Örnekler
print(is_palindrome("A man a plan a canal Panama"))  # True
print(is_palindrome("hello"))                        # False
```

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return abs(a)

# Örnekler
print(gcd(54, 24))  # 6
print(gcd(7, 13))   # 1
```

```python
def find_missing(nums, n=None):
    # eğer n verilmediyse N = len(nums)+1 varsay
    if n is None:
        n = len(nums) + 1
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

# Örnekler
print(find_missing([1,2,4,5], n=5))  # 3
# İkinci örnek 0..9 arası eksik 8: burada formül 0..N için farklı
def find_missing_zero_based(nums, n):
    # nums içinde 0..n (bir eksik) var
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(find_missing_zero_based([9,6,4,2,3,5,7,0,1], 9))  # 8
```

```python
def rotate_right(nums, k):
    if not nums:
        return nums
    k %= len(nums)
    return nums[-k:] + nums[:-k]

# Örnekler
print(rotate_right([1,2,3,4,5], 2))       # [4,5,1,2,3]
print(rotate_right([-1,-100,3,99], 2))    # [3,99,-1,-100]
```

```python
def find_median_sorted_arrays(a, b):
    # ensure a is the smaller
    if len(a) > len(b):
        a, b = b, a
    m, n = len(a), len(b)
    imin, imax, half = 0, m, (m + n + 1) // 2
    while imin <= imax:
        i = (imin + imax) // 2
        j = half - i
        if i < m and b[j-1] > a[i]:
            imin = i + 1
        elif i > 0 and a[i-1] > b[j]:
            imax = i - 1
        else:
            if i == 0:
                max_left = b[j-1]
            elif j == 0:
                max_left = a[i-1]
            else:
                max_left = max(a[i-1], b[j-1])

            if (m + n) % 2 == 1:
                return float(max_left)

            if i == m:
                min_right = b[j]
            elif j == n:
                min_right = a[i]
            else:
                min_right = min(a[i], b[j])

            return (max_left + min_right) / 2.0

# Örnekler
print(find_median_sorted_arrays([1,3], [2]))      # 2.0
print(find_median_sorted_arrays([1,2], [3,4]))    # 2.5
```

```python
def longest_palindrome(s):
    if not s:
        return ""
    start, end = 0, 0
    def expand(l, r):
        while l >= 0 and r < len(s) and s[l] == s[r]:
            l -= 1
            r += 1
        return l+1, r-1

    for i in range(len(s)):
        l1, r1 = expand(i, i)      # tek merkez
        l2, r2 = expand(i, i+1)    # çift merkez
        if r1 - l1 > end - start:
            start, end = l1, r1
        if r2 - l2 > end - start:
            start, end = l2, r2
    return s[start:end+1]

# Örnekler
print(longest_palindrome("babad"))  # "bab" veya "aba"
print(longest_palindrome("cbbd"))   # "bb"
```

```python
from collections import deque

def ladder_length(begin, end, word_list):
    word_set = set(word_list)
    if end not in word_set:
        return 0
    q = deque([(begin, 1)])
    visited = {begin}
    while q:
        word, depth = q.popleft()
        if word == end:
            return depth
        for i in range(len(word)):
            for c in 'abcdefghijklmnopqrstuvwxyz':
                if c == word[i]: 
                    continue
                new_word = word[:i] + c + word[i+1:]
                if new_word in word_set and new_word not in visited:
                    visited.add(new_word)
                    q.append((new_word, depth + 1))
    return 0

# Örnekler
print(ladder_length("hit", "cog", ["hot","dot","dog","lot","log","cog"]))  # 5
print(ladder_length("hit", "cog", ["hot","dot","dog","lot","log"]))        # 0
```

```python
def max_subarray(nums):
    max_current = max_global = nums[0]
    for x in nums[1:]:
        max_current = max(x, max_current + x)
        if max_current > max_global:
            max_global = max_current
    return max_global

# Örnekler
print(max_subarray([-2,1,-3,4,-1,2,1,-5,4]))  # 6
print(max_subarray([5,4,-1,7,8]))             # 23
```

```python
def merge_intervals(intervals):
    if not intervals:
        return []
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for cur in intervals[1:]:
        last = merged[-1]
        if cur[0] <= last[1]:  # çakışma
            last[1] = max(last[1], cur[1])
        else:
            merged.append(cur)
    return merged

# Örnekler
print(merge_intervals([[1,3],[2,6],[8,10],[15,18]]))  # [[1,6],[8,10],[15,18]]
print(merge_intervals([[1,4],[4,5]]))                 # [[1,5]]
```
